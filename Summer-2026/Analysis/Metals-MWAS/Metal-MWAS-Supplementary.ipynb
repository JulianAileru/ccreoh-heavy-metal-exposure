```python
def coalesce_metadata(df, metal_col_to_drop="metal_MWASModel"):
    id_bases = ["feature", "list_all_ids", "list_all_library", "highest_OL"]
    for base in id_bases:
        cols = [c for c in df.columns if c.startswith(f"{base}_")]
        if cols:
            df[base] = df[cols[0]]
            for c in cols[1:]:
                df[base] = df[base].fillna(df[c])
            df = df.drop(columns=cols)

    df = df.drop(columns=[c for c in df.columns if c.startswith("Unnamed:")], errors='ignore')
    df = df.drop(columns=[metal_col_to_drop], errors='ignore')

    # move the coalesced identity columns to the front
    front = [c for c in id_bases if c in df.columns]
    return df[front + [c for c in df.columns if c not in front]]


def read_outcome_sheet(path, sheet_name):
    """Return the sheet, or None if this metal/timepoint had no candidate features."""
    try:
        return pd.read_excel(path, sheet_name=sheet_name)
    except ValueError as e:
        print(f"{sheet_name}: {e}")
        return None


def merge_timepoint(mwas_df, outcome_bw, outcome_ga):
    mwas_df = mwas_df.add_suffix("_MWASModel")
    merged = mwas_df

    if outcome_bw is not None:
        outcome_bw = outcome_bw.copy()
        outcome_bw['Direction'] = np.where(outcome_bw['coef'] > 0, "Positive", 'Negative')
        outcome_bw = outcome_bw.add_suffix("_OutcomeModel_BW")
        merged = pd.merge(merged, outcome_bw, left_on="feature_MWASModel", right_on="feature_OutcomeModel_BW", how='outer')

    if outcome_ga is not None:
        outcome_ga = outcome_ga.copy()
        outcome_ga['Direction'] = np.where(outcome_ga['coef'] > 0, "Positive", 'Negative')
        outcome_ga = outcome_ga.add_suffix("_OutcomeModel_GA")
        merged = pd.merge(merged, outcome_ga, left_on="feature_MWASModel", right_on="feature_OutcomeModel_GA", how='outer')

    return coalesce_metadata(merged)


with pd.ExcelWriter("~/UNC-CH/TIMEPOINT-1-All-Models-080526.xlsx") as writer1, \
     pd.ExcelWriter("~/UNC-CH/TIMEPOINT-2-All-Models-080526.xlsx") as writer2:

    for metal in metals:
        print(metal)
        t1 = pd.read_excel(mwas_t1, sheet_name=metal)
        t2 = pd.read_excel(mwas_t2, sheet_name=metal)

        t1["Candidate"] = np.where(t1['FDR'] < 0.2, f"{metal} Candidate Feature", "Not Significant")
        t2["Candidate"] = np.where(t2['FDR'] < 0.2, f"{metal} Candidate Feature", "Not Significant")
        t1['Direction'] = np.where(t1['coef'] > 0, "Positive", 'Negative')
        t2['Direction'] = np.where(t2['coef'] > 0, "Positive", 'Negative')

        # Each timepoint is read/merged independently so a metal with zero
        # candidates at one timepoint (no outcome-model sheet for it) doesn't
        # cause the OTHER timepoint's real candidate results to be dropped.
        outcome_GA_T1 = read_outcome_sheet(gestational_age, f"{metal}-T1")
        outcome_BW_T1 = read_outcome_sheet(birthweight, f"{metal}-T1")
        outcome_GA_T2 = read_outcome_sheet(gestational_age, f"{metal}-T2")
        outcome_BW_T2 = read_outcome_sheet(birthweight, f"{metal}-T2")

        merge_t1 = merge_timepoint(t1, outcome_BW_T1, outcome_GA_T1)
        merge_t1.to_excel(writer1, sheet_name=metal)

        merge_t2 = merge_timepoint(t2, outcome_BW_T2, outcome_GA_T2)
        merge_t2.to_excel(writer2, sheet_name=metal)
```
